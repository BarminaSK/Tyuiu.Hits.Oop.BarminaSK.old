3.  Наследование

```polyglot-notebook
class Animal
{
   private string name;
   public string Name
   {
        get{return name;}
        set{name = value;}
   }
    public void MakeSound()
    {
        Console.WriteLine("Животное издаёт звук");
    }
}

class Cat : Animal
{

}

Animal myAnimal = new Animal();
Cat myCat = new Cat();

myAnimal.Name = "Шарик";
myAnimal.MakeSound();
Console.WriteLine(myAnimal.Name);

myCat.Name = "Мурзик";
myCat.MakeSound();
Console.WriteLine(myCat.Name);
```

4. Полиморфизм

```polyglot-notebook
class Animal
{
    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издаёт звук");
    }
}

class Cat : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Кот мяукает");
    }
}

Animal myAnimal = new Animal();
Cat myCat = new Cat();

myAnimal.MakeSound();
myCat.MakeSound();
```

5. Абстракция

```polyglot-notebook
using System;

abstract class Animal
{
    public abstract void MakeSound();

    public void Sleep()
    {
        Console.WriteLine("Животное спит");
    }

}

class Cat : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Кот мяучит: Мяу-Мяу!");
    }
}

Cat myCat = new Cat();

myCat.Sleep();
myCat.MakeSound();
```

6. Ассоциация (Композиция)

```polyglot-notebook
using System;

public class Heart
{
    public void Beat()
    {
        Console.WriteLine("Сердце бьётся");
    }
}

class Animal
{
    private Heart heart;

    public Animal()
    {
        this.heart = new Heart();
    }

    public void Live()
    {
        Console.WriteLine("Животное живёт");
        heart.Beat();
    }
}

Animal myAnimal = new Animal();
myAnimal.Live();
```

7. Ассоциация (Агрегация)

```polyglot-notebook
using System;

class Animal
{
    public string Name;

    public Animal(string name)
    {
        Name = name;
    }
}

class Cage
{
    public Animal Resident;

    public Cage (Animal currentAnimal)
    {
        Resident = currentAnimal;
    }
}

Animal myCat = new Animal("Мурзик");
Cage myCage = new Cage(myCat);
Console.WriteLine($"В вольере сейчас живёт: {myCage.Resident.Name}");
```
