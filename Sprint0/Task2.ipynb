Принципы SOLID

1. Принцип единственной обязанности (Single Responsibility Principle, SRP)

```polyglot-notebook
class Animal
{
    public string Name{get; set;}
    
    public void Eat()
    {
        Console.WriteLine($"{Name} ест");
    }
}

public interface IFlyable
{
    void Fly();
}

class Bird : Animal, IFlyable
{
    public void Fly()
    {
        Console.WriteLine($"Я {Name}! Я умею летать");
    }
}

Animal Wolf = new Animal();
Wolf.Name = "Волк";
Bird Sparrow = new Bird();
Sparrow.Name = "Воробей";

Wolf.Eat();
Sparrow.Eat();
Sparrow.Fly();
```

2. Принцип открытости/закрытости (Open/Closed Principle, OCP)

```polyglot-notebook
public interface IMakeSound
{
    void MakeSound();
}

class Animal : IMakeSound
{
    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издаёт звук");
    }
}

class Bird : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Птичка чирикает");
    }

    public void Fly()
    {
        Console.WriteLine("Птичка летает");
    }
}

Animal myAnimal = new Animal();
Bird myBird = new Bird();

myAnimal.MakeSound();
myBird.MakeSound();
myBird. Fly();
```

3. Принцип Liskov Substitution Principle (Принцип подстановки Лисков)

```polyglot-notebook
public interface IFlyable
{
    void Fly();
}

public interface ISwimmable
{
    void Swim();
}

public class Animal : IFlyable
{
    public string Name {get; set;}

    public void Fly()
    {
        Console.WriteLine($"{Name} умеет летать");
    } 
}

public class Fish : ISwimmable
{
    public string Name {get; set;}

    public void Swim()
    {
        Console.WriteLine($"{Name} умеет плавать");
    }
}

Animal myAnimal = new Animal();
Fish myFish = new Fish();

myAnimal.Name = "Ворона";
myFish.Name = "Треска";

myAnimal.Fly();
myFish.Swim();
```

4. Принцип разделения интерфейсов (Interface Segregation Principle, ISP)

```polyglot-notebook
public interface IMakeSound
{
    void MakeSound();
}

public interface ISleep
{
    void Sleep();
}

public interface IFly
{
    void Fly();
}

public interface ISwim
{
    void Swim();
}

public class Animal : IMakeSound, ISleep
{
    public void MakeSound()
    {
        Console.WriteLine("Животное издаёт звук");
    }

    public void Sleep()
    {
        Console.WriteLine("Животное спит");
    }
}

public class Bird : IMakeSound, ISleep, IFly
{
    public void MakeSound()
    {
        Console.WriteLine("Птичка чирикает");
    }

    public void Sleep()
    {
        Console.WriteLine("Птичка спит");
    }

    public void Fly()
    {
            Console.WriteLine("Птичка летит");
    }
}

public class Fish : ISleep, ISwim
{
    public void Sleep()
    {
        Console.WriteLine("Рыба спит");
    }

    public void Swim()
    {
            Console.WriteLine("Рыба плывёт");
    }
}

Animal myAnimal = new Animal();
Bird myBird = new Bird();
Fish myFish = new Fish();

myAnimal.MakeSound();
myAnimal.Sleep();

myBird.MakeSound();
myBird.Sleep(); 
myBird.Fly();

myFish.Sleep();
myFish.Swim();
```

5. Принцип инверсии зависимостей (Dependency Inversion Principle, DIP)

```polyglot-notebook
public interface IEnviroment
{
    void StayHere();
}

public class Forest : IEnviroment
{
    public void StayHere()
    {
        Console.WriteLine("в лесу");
    }
}

public class Ocean : IEnviroment
{
    public void StayHere()
    {
        Console.WriteLine("в океане");
    }
}

public class Animal
{
    public string Name {get; set;}
    private IEnviroment _enviroment;

    public Animal(string name, IEnviroment enviroment)
    {
        Name = name;
        _enviroment = enviroment;
    }

    public void WhereDoYouLive()
    {
        Console.Write($"{Name} живёт ");
        _enviroment.StayHere();
    }
}

IEnviroment forest = new Forest();
IEnviroment ocean = new Ocean();

Animal Wolf = new Animal("Волк", forest);
Wolf.WhereDoYouLive();

Animal Shark = new Animal("Акула", ocean);
Shark.WhereDoYouLive();
```
