```polyglot-notebook
int[] numbers = {5,8,1,3,9,6,7,2};
```

```polyglot-notebook
int min = numbers[0];
int max = numbers[0];

for (int i = 1; i < numbers.Length;i++)
{
    if (numbers[i] < min) min = numbers[i];
    if (numbers[i] > max) max = numbers[i];
}

Console.WriteLine($"1. Минимальное значение: {min}, Максимальное значение: {max}");
```

```polyglot-notebook
int sum = 0;

for (int i = 0; i < numbers.Length; i++)
{
   sum += numbers[i];
}

double average = (double)sum/numbers.Length;

Console.WriteLine($"Сумма: {sum}");
Console.WriteLine($"Среднее арифметическое: {average}");
```

```polyglot-notebook
for (int i = 0; i < numbers.Length - 1; i++)
{
    for (int j = 0; j < numbers.Length - 1 - i; j++)
    {
        if (numbers[j] > numbers[j + 1])
        {
            int current = numbers[j];
            numbers[j] = numbers[j + 1];
            numbers[j + 1] = current;
        }
    }
}

Console.WriteLine($"4. Отсортированный массив: {string.Join(", ", numbers)}");
```

```polyglot-notebook
int target = 5;
bool check = false;

for (int i = 0; i < numbers.Length; i++)
{
    if (numbers[i] == target)
    {
        check = true;
        break;
    }
}

Console.WriteLine($"5. Содержит ли массив число {target}? {(check? "Да" : "Нет")}");
```
